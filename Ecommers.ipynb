```python
import pandas as pd

df=pd.read_csv("ecommerce_sales_1000_records.csv")

print(df.head())

print(df.info())

print(df.isnull().sum())

#print(df.drop_duplicate())

df=df.drop_duplicates()
df["Date"]=pd.to_datetime(df["Date"])
```

```python
df["Total_sales"]=df["Quantity"]*df["Price"]
df
```

```python
df["Month"]=df["Date"].dt.month_name()
df
```

```python
df["Year"]=df["Date"].dt.year
df
```

```python
df["Week_days"]=df["Date"].dt.day
df
```

```python
df.groupby("Category")["Total_sales"].sum().sort_values(ascending=False)
```

```python
df.groupby("Category")["Total_sales"].sum()
```

```python
df.groupby("City")["Total_sales"].sum()
```

```python
df.groupby("Category")["Quantity"].sum()
```

```python
df.groupby("Month")["Total_sales"].sum().sort_values(ascending=False)
```

```python
import matplotlib.pyplot as plt
import pandas as pd
sales=df.groupby("Category")["Total_sales"].sum()
sales.plot(kind="bar")
plt.show()
```

```python
df.to_csv("clean_sales.csv", index=False)
```

```python
from sqlalchemy import create_engine
engine=create_engine("mysql+pymysql://root:@localhost/ecommers")

df.to_sql(
    "clean_sales",
    con=engine,
    if_exists="replace",
    index=False
)
```
